Goal here is to just write a little script to merge the proofreading excel sheet with the URLs to the IIIF images in FTP, hopefully to make proofreading a bit simpler.

```python
import pandas as pd
import numpy as np

catalog_path = 'catalog_01072025.csv'

catalog = pd.read_csv(catalog_path)
```

```python
box_16 = catalog[catalog['Box'].str.startswith('Box 16', na=False)]
```

```python
box_16.columns
```

```python
box_16_cols = box_16[['ID', 'Box', 'title', 'Document Length', 'Workflow Stage', 'Image Identifier']]
```

```python
box_16_cols
```

```python
box_16_cols['has_images'] = np.where(box_16_cols['Image Identifier'].notna(), 'Has images in Drupal', 'Does not have images in Drupal')
```

```python
box_16_cols.has_images.value_counts()
```

```python
box_16_cols['Document Length'] = box_16_cols['Document Length'].astype(int)
```

```python
box_16_cols.Box.unique()
```

```python
box_16_cols[box_16_cols['Workflow Stage'] == 'Early Access']
```

```python
box_16_cols.to_csv('box_16_cols.csv')
```
